```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")
```

```python
import pandas as pd

# Replace 'your_file.csv' with your actual file name or path
df = pd.read_csv("screen_time.csv")

# Optional: check the first few rows to confirm it loaded correctly
df.head()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Average Screen Time (hours)'], bins=10, kde=True)
plt.title("Distribution of Daily Screen Time (hours)")
plt.xlabel("Average Screen Time (hours)")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x='Age', y='Average Screen Time (hours)', data=df)
plt.title("Screen Time Distribution by Age")
plt.show()
```

```python
plt.figure(figsize=(6,5))
sns.barplot(x='Gender', y='Average Screen Time (hours)', data=df, estimator='mean', errorbar=None)
plt.title("Average Screen Time by Gender")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x='Screen Time Type', y='Average Screen Time (hours)', data=df, estimator='mean', errorbar=None)
plt.title("Average Screen Time by Activity Type")
plt.xticks(rotation=30)
plt.show()
```

```python
plt.figure(figsize=(6,5))
sns.boxplot(x='Day Type', y='Average Screen Time (hours)', data=df)
plt.title("Screen Time on Weekdays vs Weekends")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='Age', y='Average Screen Time (hours)', data=df, hue='Gender')
plt.title("Screen Time vs Age by Gender")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title("Correlation Matrix")
plt.show()
```

```python
plt.figure(figsize=(7,5))
df['Screen Time Type'].value_counts().plot(kind='pie', autopct='%1.1f%%', startangle=90)
plt.title("Distribution of Screen Time by Type")
plt.ylabel("")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x='Age', y='Average Screen Time (hours)', hue='Day Type', data=df, errorbar=None)
plt.title("Average Screen Time by Age and Day Type")
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.barplot(x='Gender', y='Average Screen Time (hours)', hue='Day Type', data=df, errorbar=None)
plt.title("Screen Time by Gender across Weekday/Weekend")
plt.show()
```

```python
plt.figure(figsize=(9,5))
sns.barplot(x='Screen Time Type', y='Average Screen Time (hours)', hue='Gender', data=df, errorbar=None)
plt.title("Activity Type Comparison by Gender")
plt.xticks(rotation=30)
plt.show()
```

```python
avg_age = df.groupby('Age')['Average Screen Time (hours)'].mean().reset_index()
plt.figure(figsize=(8,5))
sns.lineplot(x='Age', y='Average Screen Time (hours)', data=avg_age, marker='o')
plt.title("Average Screen Time Trend by Age")
plt.show()
```
