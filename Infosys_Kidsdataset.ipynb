**finding true minimum as per age group**

```python
above_limit= child_df[child_df['Exceeded_Recommended_Limit'] == True]
above_limit.sort_values(by=['Age','Avg_Daily_Screen_Time_hr'],ascending=[True,True])
```

```python
ages = sorted(child_df['Age'].unique())  # all unique ages in dataset
threshold_data = []

for age in ages:   
    max_False = within_limit.loc[(within_limit['Age'] == age), 'Avg_Daily_Screen_Time_hr'].sort_values(ascending=False)
    min_True= above_limit.loc[(above_limit['Age'] == age) ,'Avg_Daily_Screen_Time_hr'].sort_values(ascending=True)
    
    if not max_False.empty and not min_True.empty:
            threshold = (max_False.iloc[0] + min_True.iloc[0]) / 2
    else:
            threshold = None  # fallback if missing data
    
    threshold_data.append({'Age': age,'Threshold_Limit_hr': threshold})

threshold_df1 = pd.DataFrame(threshold_data)
threshold_df1
```

```python
ages = sorted(child_df['Age'].unique())  # all unique ages in dataset
Gender=child_df['Gender'].unique()
threshold_data = []

for age in ages:
    for gender in Gender:    
            max_False = within_limit.loc[(within_limit['Age'] == age) & 
                                         (within_limit['Gender']==gender), 'Avg_Daily_Screen_Time_hr'].sort_values(ascending=False)
            min_True= above_limit.loc[(above_limit['Age'] == age) & 
                                      (above_limit['Gender']==gender),'Avg_Daily_Screen_Time_hr'].sort_values(ascending=True)
    
            if not max_False.empty and not min_True.empty:
                threshold = (max_False.iloc[0] + min_True.iloc[0]) / 2
            else:
                threshold = None  # fallback if missing data
    
            threshold_data.append({'Age': age, 'Gender': gender ,'Threshold_Limit_hr': threshold})

threshold_df2 = pd.DataFrame(threshold_data)
threshold_df2
```

```python
ages = sorted(child_df['Age'].unique())
genders = child_df['Gender'].unique()
devices = child_df['Primary_Device'].unique()

threshold_data = []

for age in ages:
    for gender in genders:
        for device in devices:
            max_False = within_limit.loc[
                (within_limit['Age'] == age) &
                (within_limit['Gender'] == gender) &
                (within_limit['Primary_Device'] == device),
                'Avg_Daily_Screen_Time_hr'
            ].sort_values(ascending=False)
            
            min_True = above_limit.loc[
                (above_limit['Age'] == age) &
                (above_limit['Gender'] == gender) &
                (above_limit['Primary_Device'] == device),
                'Avg_Daily_Screen_Time_hr'
            ].sort_values(ascending=True)

            if not max_False.empty and not min_True.empty:
                threshold = (max_False.iloc[0] + min_True.iloc[0]) / 2
            else:
                threshold = None

            threshold_data.append({
                'Age': age,
                'Gender': gender,
                'Primary_Device': device,
                'Threshold_Limit_hr': threshold
            })

threshold_df3 = pd.DataFrame(threshold_data)
threshold_df3
```

```python
ages = sorted(child_df['Age'].unique())
genders = child_df['Gender'].unique()
devices = child_df['Primary_Device'].unique()
health_impact=child_df['Health_Impacts'].unique()
threshold_data = []

for age in ages:
    for gender in genders:
        for device in devices:
            for impact in health_impact:
                max_False = within_limit.loc[
                    (within_limit['Age'] == age) &
                    (within_limit['Gender'] == gender) &
                    (within_limit['Primary_Device'] == device),
                    'Avg_Daily_Screen_Time_hr'
                ].sort_values(ascending=False)
            
                min_True = above_limit.loc[
                    (above_limit['Age'] == age) &
                    (above_limit['Gender'] == gender) &
                    (above_limit['Primary_Device'] == device),
                    'Avg_Daily_Screen_Time_hr'
                ].sort_values(ascending=True)

                if not max_False.empty and not min_True.empty:
                    threshold = (max_False.iloc[0] + min_True.iloc[0]) / 2
                else:
                    threshold = None

                threshold_data.append({
                'Age': age,
                'Gender': gender,
                'Primary_Device': device,
                'Threshold_Limit_hr': threshold,
                'health_impact':impact})

threshold_df4 = pd.DataFrame(threshold_data)
threshold_df4
```

**INTEGRATING FEATURES INTO RECOMMENDATION SYSTEM**

```python
age = int(input("Enter your age (8-18) in years: "))
gender = input("Enter the gender (Male/Female): ").capitalize()
device = input("Enter your primary screen type ('Smartphone', 'Laptop', 'TV', 'Tablet'): ").capitalize()
screen_time = float(input("Enter your screen time in hours: "))

match_df = recom_df[
    (recom_df['Age'] == age) &
    (recom_df['Gender'] == gender) &
    (recom_df['Primary_Device'] == device)]    


print("\n====== RECOMMENDATION SYSTEM ======\n")
print(f'''Entered Age : {age}
Entered Gender: {gender}'
Entered Device: {device}'
Entered Screen Time : {screen_time:.2f} hours''')


if not match_df.empty:
    limit = match_df['Threshold_Limit_hr'].iloc[0]
    median=match_df['Avg_Screen_Time_hr'].iloc[0]
    
    if screen_time > limit:
        print(f"For a {age} year old {gender} child, the recommended screen time usage is {limit:.2f} hours")
        print(f"You have exceeded the recommended limit by {screen_time - limit:.2f} hours.")
    else:
        print(f"You are within the recommended healthy screen time range of {limit} hours for a {age} year old")

    diff_hr=screen_time-median
    diff_pct=(diff_hr/median)*100    

    if diff_hr > 0:
        print(f"Your usage is {diff_hr:.2f} hours ({diff_pct:.1f}%) above the average screen time "
              f"for {gender}s aged {age} using {device}.")
    elif diff_hr < 0:
        print(f"Your usage is {abs(diff_hr):.2f} hours ({abs(diff_pct):.1f}%) below the average screen time "
              f"for {gender}s aged {age} using {device}.")
    else:
        print("Your usage exactly matches the average for your group ")    
else:
    print("Sorry, no threshold found for this combination — not enough data.")

print("\n===================================")
```
