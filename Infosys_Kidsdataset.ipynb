```python
ages = sorted(child_df['Age'].unique())  # all unique ages in dataset
threshold_data = []

for age in ages:   
    max_False = within_limit.loc[(within_limit['Age'] == age), 'Avg_Daily_Screen_Time_hr'].sort_values(ascending=False)
    min_True= above_limit.loc[(above_limit['Age'] == age) ,'Avg_Daily_Screen_Time_hr'].sort_values(ascending=True)
    
    if not max_False.empty and not min_True.empty:
            threshold = (max_False.iloc[0] + min_True.iloc[0]) / 2
    else:
            threshold = None  # fallback if missing data
    
    threshold_data.append({'Age': age,'Threshold_Limit_hr': threshold})

threshold_df1 = pd.DataFrame(threshold_data)
threshold_df1
```

```python
ages = sorted(child_df['Age'].unique())  # all unique ages in dataset
Gender=child_df['Gender'].unique()
threshold_data = []

for age in ages:
    for gender in Gender:    
            max_False = within_limit.loc[(within_limit['Age'] == age) & 
                                         (within_limit['Gender']==gender), 'Avg_Daily_Screen_Time_hr'].sort_values(ascending=False)
            min_True= above_limit.loc[(above_limit['Age'] == age) & 
                                      (above_limit['Gender']==gender),'Avg_Daily_Screen_Time_hr'].sort_values(ascending=True)
    
            if not max_False.empty and not min_True.empty:
                threshold = (max_False.iloc[0] + min_True.iloc[0]) / 2
            else:
                threshold = None  # fallback if missing data
    
            threshold_data.append({'Age': age, 'Gender': gender ,'Threshold_Limit_hr': threshold})

threshold_df2 = pd.DataFrame(threshold_data)
threshold_df2
```

```python
ages = sorted(child_df['Age'].unique())
genders = child_df['Gender'].unique()
devices = child_df['Primary_Device'].unique()

threshold_data = []

for age in ages:
    for gender in genders:
        for device in devices:
            max_False = within_limit.loc[
                (within_limit['Age'] == age) &
                (within_limit['Gender'] == gender) &
                (within_limit['Primary_Device'] == device),
                'Avg_Daily_Screen_Time_hr'
            ].sort_values(ascending=False)
            
            min_True = above_limit.loc[
                (above_limit['Age'] == age) &
                (above_limit['Gender'] == gender) &
                (above_limit['Primary_Device'] == device),
                'Avg_Daily_Screen_Time_hr'
            ].sort_values(ascending=True)

            if not max_False.empty and not min_True.empty:
                threshold = (max_False.iloc[0] + min_True.iloc[0]) / 2
            else:
                threshold = None

            threshold_data.append({
                'Age': age,
                'Gender': gender,
                'Primary_Device': device,
                'Threshold_Limit_hr': threshold
            })

threshold_df3 = pd.DataFrame(threshold_data)
threshold_df3
```

```python
ages = sorted(child_df['Age'].unique())
genders = child_df['Gender'].unique()
devices = child_df['Primary_Device'].unique()
Health_Impacts=child_df['Health_Impacts'].unique()
threshold_data = []

for age in ages:
    for gender in genders:
        for device in devices:
            for impact in Health_Impacts:
                max_False = within_limit.loc[
                    (within_limit['Age'] == age) &
                    (within_limit['Gender'] == gender) &
                    (within_limit['Primary_Device'] == device),
                    'Avg_Daily_Screen_Time_hr'
                ].sort_values(ascending=False)
            
                min_True = above_limit.loc[
                    (above_limit['Age'] == age) &
                    (above_limit['Gender'] == gender) &
                    (above_limit['Primary_Device'] == device),
                    'Avg_Daily_Screen_Time_hr'
                ].sort_values(ascending=True)

                if not max_False.empty and not min_True.empty:
                    threshold = (max_False.iloc[0] + min_True.iloc[0]) / 2
                else:
                    threshold = None

                threshold_data.append({
                'Age': age,
                'Gender': gender,
                'Primary_Device': device,
                'Health_Impacts':impact,
                'Threshold_Limit_hr': threshold}
                )

threshold_df4 = pd.DataFrame(threshold_data)
threshold_df4=threshold_df4.round(2)
threshold_df4
```

```python
age = int(input("Enter your age (8-18) in years: "))
gender = input("Enter the gender (Male/Female): ").capitalize()
device = input("Enter your primary screen type ('Smartphone', 'Laptop', 'TV', 'Tablet'): ").capitalize()
edu_time = float(input("Enter your educational screen time (in hours): "))
rec_time = float(input("Enter your recreational screen time (in hours): "))
screen_time = edu_time + rec_time

# making match_df from recomdf2 by filtering the data in it with user inputs
match_df = recom_df2[
    (recom_df2['Age'] == age) &
    (recom_df2['Gender'] == gender) &
    (recom_df2['Primary_Device'] == device)
]

print("\n================================= RECOMMENDATION SYSTEM ===================================\n")
print(f'''Entered Age : {age}
Entered Gender: {gender}
Entered Device: {device}
Entered Edu. Screen time: {edu_time}
Entered Rec. Screen time: {rec_time}
Total Screen Time : {screen_time:.2f} hours''')

if not match_df.empty:
    limit = match_df['Threshold_Limit_hr'].iloc[0]
    avg_detailed = match_df['Avg_screen_time_per_criterias'].iloc[0]
    avg_age_only = match_df['Avg_Daily_Screen_Time_per_age_only'].iloc[0]

    # --------------- Recommendation logic ---------------
    #checking whether the user's screen time is more than the thershold for that for that combination of age, gender and device  
    if screen_time > limit:
        print(f"\nFor a {age}-year-old {gender} child, the recommended screen time usage is {limit:.2f} hours.")
        print(f"You have exceeded the recommended limit by {screen_time - limit:.2f} hours.")
    else:
        print(f"\nYou are within the recommended healthy screen time range of {limit:.2f} hours for a {age}-year-old.")

    # certain user's screen time against their criterias group statistics
    diff_hr = screen_time - avg_detailed
    diff_pct = (diff_hr / avg_detailed) * 100 if avg_detailed != 0 else 0

    if diff_hr > 0:
        print(f"\nYour usage is {diff_hr:.2f} hours ({diff_pct:.1f}%) above the average screen time "
              f"for {gender}s aged {age} using {device}.")
    elif diff_hr < 0:
        print(f"\nYour usage is {abs(diff_hr):.2f} hours ({abs(diff_pct):.1f}%) below the average screen time "
              f"for {gender}s aged {age} using {device}.")
    else:
        print("\nYour usage exactly matches the average for your group.")

    # certain user's screen time against overall age group only statistics
    diff_age_hr = screen_time - avg_age_only
    diff_age_pct = (diff_age_hr / avg_age_only) * 100

    if diff_age_hr > 0:
        print(f"\nCompared to the overall average for a child aged {age}, your usage is {diff_age_hr:.2f} hours ({diff_age_pct:.1f}%) higher.")
    elif diff_age_hr < 0:
        print(f"\nCompared to the overall average for a child aged {age}, your usage is {abs(diff_age_hr):.2f} hours ({abs(diff_age_pct):.1f}%) lower.")
    else:
        print(f"\nYour usage exactly matches the overall average screen time for age {age}.")
    
    # ===================== EDUCATIONAL & RECREATIONAL TIME COMPARISON =====================
    avg_edu_age = match_df['Avg_Educational_Screen_time'].iloc[0]
    avg_rec_age = match_df['Avg_Recreational_Screen_time'].iloc[0]

    # --- Educational screen time comparison ---
    edu_diff_hr = edu_time - avg_edu_age
    edu_diff_pct = (edu_diff_hr / avg_edu_age) * 100 if avg_edu_age != 0 else 0

    if edu_diff_hr > 0:
        print(f"\nYour educational screen time is {edu_diff_hr:.2f} hours ({edu_diff_pct:.1f}%) higher "
            f"than the average for {age}-year-old {gender}s using {device}.")
        print("You spend more time studying or using screens for learning than most children in your group.")
    elif edu_diff_hr < 0:
        print(f"\nYour educational screen time is {abs(edu_diff_hr):.2f} hours ({abs(edu_diff_pct):.1f}%) lower "
            f"than the average for {age}-year-old {gender}s using {device}.")
        print("You spend less time on educational activities compared to others in your group. Consider increasing your focus on learning-based use.")
    else:
        print(f"\nYour educational screen time matches the average of {avg_edu_age:.2f} hours "
            f"for {age}-year-old {gender}s using {device}.")
        print("You are aligned with how much time other children in your group spend on educational use.")

    # --- Recreational screen time comparison ---
    rec_diff_hr = rec_time - avg_rec_age
    rec_diff_pct = (rec_diff_hr / avg_rec_age) * 100 if avg_rec_age != 0 else 0

    if rec_diff_hr > 0:
        print(f"\nYour recreational screen time is {rec_diff_hr:.2f} hours ({rec_diff_pct:.1f}%) higher "
            f"than the average for {age}-year-old {gender}s using {device}.")
        print("You spend more time on entertainment or leisure than others in your group. Consider reducing this to maintain balance.")
    elif rec_diff_hr < 0:
        print(f"\nYour recreational screen time is {abs(rec_diff_hr):.2f} hours ({abs(rec_diff_pct):.1f}%) lower "
            f"than the average for {age}-year-old {gender}s using {device}.")
        print("You engage in less recreational screen use than others in your group, indicating healthy screen habits.")
    else:
        print(f"\nYour recreational screen time matches the average of {avg_rec_age:.2f} hours "
            f"for {age}-year-old {gender}s using {device}.")
        print("Your entertainment habits are consistent with your group’s average.")

    #calculating the educational to recreational ratio of user
    edu_to_rec_ratio_ser = recom_df2[(recom_df2['Age'] == age)]['Educational_to_Recreational_Ratio']
    edu_to_rec_ratio_ser2 = edu_to_rec_ratio_ser.iloc[0]
    
    if rec_time !=0:
        edu_to_rec_ratio=edu_time/rec_time
        print(f"\nYour Educational to Recreational screen time ratio is : {edu_to_rec_ratio:.3f}")
        print(f"The Average of educational to recreational for a {age} year old is {edu_to_rec_ratio_ser2:.3f}")
    else:
        edu_to_rec_ratio = float('inf')
        print(f"\nYour Educational to Recreational screen time ratio is : Infinite (no recreational time)")   
        print(f"The Average of educational to recreational for a {age} year old is {edu_to_rec_ratio_ser2:.3f}")  
    
    if edu_to_rec_ratio != float('inf'):
        diff_ratio = edu_to_rec_ratio - edu_to_rec_ratio_ser2   #comparing the user ratio to the avg ratio for similar age group
        if diff_ratio > 0:
            print(f"Your ratio is {diff_ratio:.3f} higher than your age group's average.")
            print("This means you spend *more time on educational activities* compared to most children your age.")
        elif diff_ratio < 0:
            print(f"Your ratio is {abs(diff_ratio):.3f} lower than your age group's average.")
            print("This means your screen time is *more focused on recreation or entertainment* than the average for your age group.")
        else:
            print("Your ratio matches your age group's average exactly.")
            print("You have a similar balance between educational and recreational screen use as your peers.")
    else:
         print(f"Your Educational to Recreational screen time ratio is : ∞ (since you reported no recreational screen time)")

    # making group_health dataframe from recom_df2 by filtering the data in it with user inputs and then selecting 'Health_Impacts' column
    group_health = recom_df2[
            (recom_df2['Age'] == age) &
            (recom_df2['Gender'] == gender) &
            (recom_df2['Primary_Device'] == device)]['Health_Impacts']

    if group_health.empty:
            print("\nHealth insights: not enough data for this group.")
    else:
            health_counts = group_health.value_counts()         #provides a series with health impact entries as index and their count as values
            top_impacts = health_counts.head(3)

    #extracting common health impacts among similar users
    print("\nMost common health impacts among similar users:")
    for impact, cnt in top_impacts.items():
                print(f"- {impact}: {cnt} users")

            
    top_list = list(top_impacts.index)      #provides a list of health impact entries
    if screen_time > limit:
                impacts_str = ", ".join(top_list[:-1]) + f", and {top_list[-1]}" if len(top_list) > 1 else top_list[0]  
                print(f"\nExceeding the screen time limit is often linked to {impacts_str} among similar users.")
    else:
        if 'None' in top_list:
            print("\nGood news! Most users with similar habits report 'None' as their main health impact.")
    
#Risk analysis based on the difference between screen time and threshold limit
    exceed_pct = ((screen_time - limit) / limit) * 100
    if exceed_pct <= 0:
        risk = " Healthy"
    elif exceed_pct <= 20:
        risk = " Mild Risk"
    elif exceed_pct <= 50:
        risk = " High Risk"
    else:
        risk = " Severe Risk"

    print(f"\nRisk Level: {risk}")     

# Device recommendation based on least health impacts (by Age only)
user_age_devices = Health_Impacts_count_df[Health_Impacts_count_df['Age'] == age]

if not user_age_devices.empty:
    # Find the device with the fewest reported health impacts
    best_device_row = user_age_devices.loc[user_age_devices['Health_Impacts_count'].idxmin()]
    best_device = best_device_row['Primary_Device']
    best_device_count = int(best_device_row['Health_Impacts_count'])

    print("\nDevice Recommendation:")
    print(f"For a {age}-year-old, the device associated with the fewest reported health issues is the '{best_device}' "
          f"({best_device_count} total health impact cases among peers).")

    if device != best_device:
        print(f"Consider switching from '{device}' to '{best_device}' for potentially healthier screen usage.")
    else:
        print(f"Good choice! '{device}' already has the lowest number of health impacts for your age group.")
else:
    print("\nNo sufficient data to recommend a device for your age group.")

print("\n=============================================================================================")
```
