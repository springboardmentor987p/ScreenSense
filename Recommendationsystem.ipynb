```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from datetime import datetime, timedelta

# ----------------------------------------------------------
# 1) LOAD DATA
# ----------------------------------------------------------

# Try loading real dataset
try:
    df = pd.read_csv("screen_usage.csv", parse_dates=["datetime"])
except:
    # Create synthetic data
    np.random.seed(42)
    n = 8000
    start = datetime(2024, 1, 1)

    ages = np.random.randint(10, 70, n)
    genders = np.random.choice(["Male", "Female", "Other"], n)
    devices = np.random.choice(["Mobile", "Tablet", "Laptop"], n, p=[0.7, 0.15, 0.15])
    categories = np.random.choice(
        ["Social", "Video", "Education", "Gaming", "Productivity", "Health"], n
    )
    dates = [start + timedelta(days=np.random.randint(0, 300)) for _ in range(n)]
    mins = np.random.randint(5, 300, n)

    df = pd.DataFrame({
        "user_id": np.random.randint(100, 200, n),
        "age": ages,
        "gender": genders,
        "device_type": devices,
        "datetime": dates,
        "app_category": categories,
        "screen_minutes": mins
    })

# Extra fields
df["date"] = df["datetime"].dt.date

# Age band function
def age_band(a):
    if a < 13:
        return "0-12"
    elif a < 18:
        return "13-17"
    elif a < 25:
        return "18-24"
    elif a < 35:
        return "25-34"
    elif a < 45:
        return "35-44"
    elif a < 60:
        return "45-59"
    else:
        return "60+"

df["age_band"] = df["age"].apply(age_band)
df["month"] = pd.to_datetime(df["date"]).dt.to_period("M").dt.to_timestamp()
df["cohort"] = df["age_band"] + " | " + df["device_type"]

# ----------------------------------------------------------
# 2) COHORT ANALYSIS
# ----------------------------------------------------------

cohort = df.groupby(["age_band", "device_type"]).agg(
    users=("user_id", "nunique"),
    sessions=("user_id", "count"),
    avg_minutes=("screen_minutes", "mean")
).reset_index()

print("\nTOP COHORTS BY SCREEN TIME\n")
print(cohort.sort_values("avg_minutes", ascending=False).head())

# ----------------------------------------------------------
# 3) COHORT × MONTH HEATMAP
# ----------------------------------------------------------

pivot = df.groupby(["cohort", "month"]).screen_minutes.mean().unstack(fill_value=0)

plt.figure(figsize=(12, 5))
plt.imshow(pivot, aspect="auto")
plt.xticks(range(len(pivot.columns)), pivot.columns, rotation=45)
plt.yticks(range(len(pivot.index)), pivot.index)
plt.colorbar()
```

```python
import pandas as pd

# -----------------------------------------------------------
# LOAD YOUR FEATURE DICTIONARY FILE
# -----------------------------------------------------------
df = pd.read_csv("Feature_Dictionary.csv")

# -----------------------------------------------------------
# AGE BAND CLASSIFIER
# -----------------------------------------------------------
def classify_age(age):
    if 8 <= age <= 11:
        return "8-11"
    elif 12 <= age <= 15:
        return "12-15"
    elif 16 <= age <= 18:
        return "16-18"
    else:
        return None

# -----------------------------------------------------------
# RECOMMENDATION LOGIC
# -----------------------------------------------------------
def recommend_device(age, screen_time):
    age_band = classify_age(age)

    if age_band is None:
        return "Age not in dataset range (8–18)."

    # Filter for kids in same age band
    similar = df[df["age_band"] == age_band].copy()

    # Compute difference between user’s screen time and data
    similar["diff"] = (similar["avg_daily_screen_time_hr"] - screen_time).abs()

    # Top 10 closest cases
    closest = similar.sort_values("diff").head(10)

    # Prefer healthy (no health issue)
    low_risk = closest[closest["has_health_issue"] == 0]

    # If low-risk data exists
    if len(low_risk) > 0:
        recommended = low_risk["primary_device"].mode()[0]
    else:
        # fallback: overall closest match
        recommended = closest["primary_device"].mode()[0]

    return recommended

# -----------------------------------------------------------
# CONSOLE APPLICATION
# -----------------------------------------------------------

print("===== Kids Screen Time Recommendation System =====")

try:
    age = int(input("Enter child's age (8–18): "))
    screen_time = float(input("Enter average daily screen time (in hours): "))

    device = recommend_device(age, screen_time)

    print("\nRecommended Device for This Child:")
    print("👉", device)

except Exception as e:
    print("Error:", e)
```

```python
def risk_dictionary(level):
    if level == "low":
        return {
            "risk": "Low Risk",
            "advice": [
                "Screen time is within a healthy range.",
                "Maintain a balanced routine of study, play and rest.",
                "Continue monitoring screen habits."
            ]
        }

    elif level == "medium":
        return {
            "risk": "Medium Risk",
            "advice": [
                "Reduce screen time by 30 to 60 minutes.",
                "Encourage at least one hour of outdoor activity.",
                "Avoid screen use before bedtime.",
                "Promote healthy breaks every 30 minutes."
            ]
        }

    elif level == "high":
        return {
            "risk": "High Risk",
            "advice": [
                "Immediate reduction in screen time is recommended.",
                "Create a strict digital schedule.",
                "Encourage non screen hobbies like sports, music or reading.",
                "Seek professional help if behaviour or sleep patterns are affected."
            ]
        }

    else:
        return {
            "risk": "Unknown",
            "advice": [
                "Invalid input received.",
                "Please enter correct screen time range."
            ]
        }

# ---------------- CALLING THE FUNCTION ----------------

level = input("Enter risk level (low, medium, high): ")

result = risk_dictionary(level)

print("Risk:", result["risk"])
print("Advice:")
for item in result["advice"]:
    print("-", item)
```
