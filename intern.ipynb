```python
avg_screen_time_gender = df.groupby('Gender')['Average Screen Time (hours)'].mean()
plt.figure(figsize=(6,4))
plt.bar(avg_screen_time_gender.index, avg_screen_time_gender.values, color=['skyblue', 'green','yellow'])
plt.title('Average Screen Time by Gender')
plt.xlabel('Gender')
plt.ylabel('Average Screen Time (hours)')
plt.show()
```

```python
avg_time_type = df.groupby('Screen Time Type')['Average Screen Time (hours)'].mean()
print(avg_time_type)
plt.figure(figsize=(6,4))
colors = ['skyblue', 'salmon', 'lightgreen']
plt.bar(avg_time_type.index, avg_time_type.values, color=colors)
plt.title('Average Screen Time by Type')
plt.xlabel('Screen Time Type')
plt.ylabel('Average Screen Time (hours)')
for i, v in enumerate(avg_time_type.values):
    plt.text(i, v + 0.02, f"{v:.2f}", ha='center', fontweight='bold')
plt.show()
```

```python
avg_time_day = df.groupby('Day Type')['Average Screen Time (hours)'].mean()
print(avg_time_day)
plt.figure(figsize=(6,6))
colors = ['skyblue', 'salmon']
plt.pie(avg_time_day.values, labels=avg_time_day.index, autopct='%1.1f%%', colors=colors, startangle=90, explode=(0.05, 0.05))
plt.title('Average Screen Time: Weekday vs Weekend')
plt.show()
```

```python
df['Gender'] = df['Gender'].astype('category')
df['Screen Time Type'] = df['Screen Time Type'].astype('category')
df['Day Type'] = df['Day Type'].astype('category')
plt.figure(figsize=(8,5))
plt.hist(df['Average Screen Time (hours)'], bins=20, edgecolor='black')
plt.title('Distribution of Average Screen Time (hours)')
plt.xlabel('Average Screen Time (hours)')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.4)
plt.show()
```

```python
import pandas as pd

print("Paste your file path below and press ENTER")

file_path = input()

print("\nYou entered:", file_path)

# Load CSV or Excel automatically
if file_path.endswith(".csv"):
    df = pd.read_csv(file_path)
else:
    df = pd.read_excel(file_path)

display(df.head())
print("\nDataset shape:", df.shape)

```

```python
df = df.drop_duplicates()
df = df.dropna(how='all')

# Fill missing values
num_cols = df.select_dtypes(include=[np.number]).columns
cat_cols = df.select_dtypes(include='object').columns

df[num_cols] = df[num_cols].fillna(df[num_cols].median())
df[cat_cols] = df[cat_cols].fillna(df[cat_cols].mode().iloc[0])

display(df.head())
```

```python
df = df.drop_duplicates()
df = df.dropna(how='all')

df.rename(columns={
    "Avg_Daily_Screen_Time_hr": "Total_ScreenTime",
    "Educational_to_Recreational_Ratio": "EduRec_Ratio",
    "Primary_Device": "Device",
    "Health_Impacts": "Health"
}, inplace=True)

df.head()
```

```python
def classify_risk(total):
    if total <= 2:
        return "Healthy"
    elif 2 < total <= 4:
        return "Moderate"
    elif 4 < total <= 6:
        return "High"
    else:
        return "Very High"

df["Risk_Level"] = df["Total_ScreenTime"].apply(classify_risk)
```

```python
def recommend(row):
    total = row["Total_ScreenTime"]

    if total <= 2:
        return "Great! Screen habits are healthy."
    elif 2 < total <= 4:
        return "Reduce some non-essential screen time."
    elif 4 < total <= 6:
        return "Take frequent breaks. Reduce long TV/mobile sessions."
    else:
        return "Very high! Reduce screen time urgently."

df["Recommendation"] = df.apply(recommend, axis=1)
```
