```python
import matplotlib.pyplot as plt

df.groupby('Day Type')['Average Screen Time (hours)'].mean().plot(kind='bar')
plt.title('Average Screen Time by Day Type')
plt.ylabel('Average Screen Time (hours)')
plt.show()
```

```python
df.groupby('Gender')['Average Screen Time (hours)'].mean().plot(kind='bar', color=['skyblue', 'lightcoral'])
plt.title('Average Screen Time by Gender')
plt.ylabel('Average Screen Time (hours)')
plt.show()
```

```python
df.groupby('Screen Time Type')['Average Screen Time (hours)'].mean().plot(kind='bar', color='orange')
plt.title('Average Screen Time by Type (e.g., Social, Entertainment, etc.)')
plt.ylabel('Average Screen Time (hours)')
plt.show()
```
