```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load your dataset
df = pd.read_csv("screen_time.csv")   # <-- make sure the file name is correct

df.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

pivot_data = df.pivot_table(
    values='Average Screen Time (hours)',
    index='Age',
    columns='Screen Time Type',
    aggfunc='mean'
)

plt.figure(figsize=(10,6))
sns.heatmap(pivot_data, annot=True, cmap='YlGnBu')
plt.title("Average Screen Time by Age Group and Screen Time Type")
plt.show()
```

```python
import pandas as pd

grouped = df.groupby(['Gender', 'Day Type'])['Average Screen Time (hours)'].mean().unstack()

grouped.plot(kind='bar', stacked=True, figsize=(8,6))
plt.title("Screen Time by Gender Across Day Types")
plt.ylabel("Average Daily Hours")
plt.show()
```

```python
df.groupby(['Age', 'Screen Time Type'])['Average Screen Time (hours)'].mean().sort_values(ascending=False).head(5)
```

```python
df.groupby('Day Type')['Average Screen Time (hours)'].mean().plot(kind='bar', figsize=(6,5))
plt.title("Weekday vs Weekend Screen Time")
plt.ylabel("Average Hours")
plt.show()
```

```python
df.groupby('Age')['Average Screen Time (hours)'].mean().plot(kind='line', marker='o', figsize=(7,5))
plt.title("Screen Time Trend Across Age Groups")
plt.ylabel("Average Hours")
plt.show()
```

```python
df.groupby(['Gender', 'Screen Time Type'])['Average Screen Time (hours)'].mean().unstack().plot(
    kind='bar', figsize=(10,6)
)
plt.title("Screen Time Type Comparison by Gender")
plt.ylabel("Hours")
plt.show()
```
