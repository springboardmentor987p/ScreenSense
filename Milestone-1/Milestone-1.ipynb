## 3. Visualizations

### 3.1 Line Chart: Avg Daily Screen Time by Age

```python
avg_screen_time_by_age = df.groupby('Age')['Avg_Daily_Screen_Time_hr'].mean()
plt.figure(figsize=(8,5))
plt.plot(avg_screen_time_by_age.index, avg_screen_time_by_age.values, 
         marker='o', color='skyblue', linestyle='-')
plt.title('Average Daily Screen Time by Age', fontsize=14, fontweight='bold')
plt.xlabel('Age')
plt.ylabel('Average Screen Time (hours)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
for x, y in zip(avg_screen_time_by_age.index, avg_screen_time_by_age.values):
    plt.text(x, y + 0.01, f"{y:.2f}", ha='center', va='bottom', fontsize=10, fontweight='normal')
plt.tight_layout()
plt.show()
```

### 3.2 Bar Chart: Number of Kids Exceeding Recommended Screen Time

```python
counts = df['Exceeded_Recommended_Limit'].value_counts()
plt.bar(counts.index.astype(str), counts.values, color=['#77DD77', '#F88379'])
plt.title('Kids Exceeding Recommended Screen Time')
plt.xlabel('Exceeded Recommended Limit')
plt.ylabel('Count')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

### 3.3 Multi-Line Chart: Average Daily Screen Time by Age and Gender

```python
avg_by_age_gender = df.groupby(['Age', 'Gender'])['Avg_Daily_Screen_Time_hr'].mean().unstack()
ages = avg_by_age_gender.index
genders = avg_by_age_gender.columns
bar_width = 0.35
x = np.arange(len(ages))  # the label locations
plt.figure(figsize=(10, 6))
for i, (gender, color) in enumerate(zip(genders, ['#4C72B0', '#E76F51'])):
    bars = plt.bar(x + i*bar_width, avg_by_age_gender[gender], 
                   width=bar_width, color=color, label=gender, alpha=0.85, edgecolor='black')
    for bar in bars:
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2, height + 0.05,
                 f"{height:.2f}", ha='center', va='bottom', fontweight='normal', fontsize=10)
plt.title('Average Daily Screen Time by Age and Gender', fontsize=14, fontweight='bold')
plt.xlabel('Age', fontsize=12)
plt.ylabel('Average Screen Time (hours)', fontsize=12)
plt.xticks(x + bar_width / 2, ages)
plt.legend(title='Gender')
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

### 3.4 Stacked Bar Chart: Health Impacts by Location Type

```python
df_exploded = (df.assign(Health_Impacts=df['Health_Impacts'].fillna('None')
                         .str.split(', '))
                 .explode('Health_Impacts')
                 .assign(Health_Impacts=lambda x: x['Health_Impacts'].str.strip()))

pivot = (df_exploded.groupby(['Urban_or_Rural','Health_Impacts'])
                 .size()
                 .unstack(fill_value=0))
pivot.plot(kind='bar', stacked=True, figsize=(8,5),
           color=['#FFA600','#4C72B0','#55A868','#E76F51','#8172B3'])
plt.title('Health Impacts by Urban vs Rural', fontsize=14, fontweight='bold')
plt.xlabel('Location', fontsize=12)
plt.ylabel('Count of Impacts', fontsize=12)
plt.legend(title='Impact', bbox_to_anchor=(1.02,1), loc='upper left')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

### 3.5 Donut Chart: Percentage of Participants by Location Type

```python
urban_rural_counts = df['Urban_or_Rural'].value_counts()
plt.figure(figsize=(6,6))
wedges, texts, autotexts = plt.pie(urban_rural_counts, autopct='%1.1f%%', startangle=90,
                                   colors=['#4C72B0', '#55A868'], wedgeprops=dict(width=0.3))
plt.legend(wedges, urban_rural_counts.index, title="Location", loc="center left", bbox_to_anchor=(1, 0, 0.5, 1))
plt.title('Participants: Urban vs Rural (Donut Chart)')
plt.show()
```

### 3.6 Pie Chart: Primary Devices Usage Distribution

```python
device_counts = df['Primary_Device'].value_counts()
plt.figure(figsize=(7,7))
device_counts.plot(kind='pie', autopct='%1.1f%%', startangle=90, 
                   colors=['#21565b', '#56d3d0', '#b2e1db', '#ffe670'])
plt.title('Distribution of Primary Devices Used')
plt.ylabel('')
plt.show()
```

### 3.7 Histogram: Ratio Of Educational and Recreational values by Frequency

```python
fig, ax = plt.subplots(figsize=(10, 6))
ax.hist(
    df['Educational_to_Recreational_Ratio'],  
    bins=20,                                  
    color='#FFD580',                         
    edgecolor='black',                        
    alpha=0.7                                 
)
ax.set_title('Distribution of Educational to Recreational Ratio', fontsize=14, fontweight='bold')
ax.set_xlabel('Educational to Recreational Ratio', fontsize=12)
ax.set_ylabel('Frequency', fontsize=12)
ax.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

### 3.8 Horizontal Grouped Chart: Average Screen Time by Gender and Device Type

```python
gender_device_data = df.groupby(['Gender', 'Primary_Device'])['Avg_Daily_Screen_Time_hr'].mean().unstack()
fig, ax = plt.subplots(figsize=(9, 5))
gender_device_data.T.plot(
    kind='barh', 
    ax=ax, 
    width=0.7,
    color=['#FF6B9D', '#4ECDC4'],
    edgecolor='black',
    linewidth=0.8
)
ax.set_title('Average Daily Screen Time by Gender and Device Type', 
             fontsize=16, fontweight='bold', pad=20)
ax.set_xlabel('Average Screen Time (hours)', fontsize=12)
ax.set_ylabel('Primary Device', fontsize=12)
ax.legend(title='Gender', fontsize=11)
ax.grid(axis='x', alpha=0.3, linestyle='--')
for container in ax.containers:
    ax.bar_label(container, fmt='%.2f', padding=3, fontsize=9)
plt.tight_layout()
plt.show()
```
